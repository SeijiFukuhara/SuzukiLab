```python
import pandas as pd
import numpy as np
from scipy.ndimage import uniform_filter1d
```

```python
# CSVファイルを読み込み
df = pd.read_csv(r'C:\\Users\\seiji\\VSCodeFolder\\SuzukiLab\data\\sample_phase.csv')
```

```python
df_uniformed = uniform_filter1d(df, size=1, axis=0, mode='nearest')
```

```python
print(np.array_equal(df, df_uniformed))  # Trueならば、元のデータと同じ
```

```python
print(np.allclose(df,df_uniformed, atol =  1e-8, rtol = 1e-5))
```
