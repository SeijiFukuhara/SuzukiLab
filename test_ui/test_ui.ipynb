```python
twolist_array = np.array([[1, 2, 3],
                     [4, 5, 6]])
print(twolist_array)
num_functions = 4
```

```python
# twolist_array を n 回複製して、新たな配列を作成
# まず、twolist_array に次元を追加 (1, 2, 3)
twolist_expanded = twolist_array[np.newaxis, :, :]

# tile で n 回繰り返す
n =5
result_array = np.tile(twolist_expanded, (n, 1, 1))
print(result_array)
```

```python
phase_full = [twolist_array for _ in range(num_functions)]  # 4つの元の配列を作成
print(phase_full)
```
